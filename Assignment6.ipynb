```python
#2
# Global variable
total_amount = 0

def calculate_bill():
    global total_amount   # declaring global variable inside function
    
    subtotal = 2420
    tax = 121.0
    discount = 242.0
    payment_method = "UPI"
    
    total_amount = subtotal + tax - discount
    
    print("Subtotal:", subtotal)
    print("Tax:", tax)
    print("Discount:", discount)
    print("Total:", total_amount)
    print("Payment:", payment_method)

# Function call
calculate_bill()

# Accessing global variable outside the function
print("Global Total Amount:", total_amount)

```

```python
#3
# Global variable
counter = 0

def increment():
    global counter
    counter += 1
    print("After increment, counter =", counter)

def decrement():
    global counter
    counter -= 1
    print("After decrement, counter =", counter)

def reset():
    global counter
    counter = 0
    print("After reset, counter =", counter)

# Function calls
increment()
increment()
decrement()
reset()

# Final value of global variable
print("Final counter value =", counter)
```

```python
#4
# Global variable to store cumulative sales
total_sales = 0

def add_online_sales(amount):
    global total_sales
    total_sales += amount
    print("Online sale added:", amount)
    print("Total sales so far:", total_sales)

def add_store_sales(amount):
    global total_sales
    total_sales += amount
    print("Store sale added:", amount)
    print("Total sales so far:", total_sales)

# Function calls
add_online_sales(1500)
add_store_sales(2000)
add_online_sales(500)
add_store_sales(1000)

# Final total sales
print("Final cumulative total sales:", total_sales)
```

```python
#6
def outer_function():
    count = 0   # Enclosing variable

    def inner_function():
        nonlocal count   # Refers to enclosing variable
        count += 1
        print("Inner function count:", count)

    inner_function()
    inner_function()
    inner_function()

    print("Final count in outer function:", count)

# Function call
outer_function()
```

```python
#7
# Global variable
global_var = "I am a GLOBAL variable"

def outer_function():
    enclosing_var = "I am an ENCLOSING variable"

    def middle_function():
        middle_var = "I am a MIDDLE (local to middle) variable"

        def inner_function():
            local_var = "I am a LOCAL variable"
            
            print(local_var)        # Local scope
            print(middle_var)       # Enclosing scope (middle)
            print(enclosing_var)    # Enclosing scope (outer)
            print(global_var)       # Global scope

        inner_function()

    middle_function()

# Function call
outer_function()
```

```python
#12

def calculate_marks():
    # Taking input in outer function
    marks1 = int(input("Enter marks for Subject 1: "))
    marks2 = int(input("Enter marks for Subject 2: "))
    marks3 = int(input("Enter marks for Subject 3: "))

    def total_marks():
        # Inner function for computation
        total = marks1 + marks2 + marks3
        return total

    # Calling inner function
    result = total_marks()
    print("Total Marks:", result)

# Function call
calculate_marks()
```

```python
#13
def counter_function():
    count = 0   # Enclosing variable

    def increment():
        nonlocal count   # Modify enclosing variable
        count += 1
        return count

    return increment

# Create counter
counter = counter_function()

# Function calls
print(counter())
print(counter())
print(counter())
print(counter())
```

```python
#14
def multiplier(factor):
    # Enclosing variable: factor

    def multiply(number):
        return number * factor   # factor is remembered

    return multiply

# Create multiplier functions
double = multiplier(2)
triple = multiplier(3)

# Using the closures
print("Double of 5:", double(5))
print("Double of 10:", double(10))

print("Triple of 5:", triple(5))
print("Triple of 10:", triple(10))
```

```python
#15
def calculate_sum(*numbers):
    total = 0
    for num in numbers:
        total += num
    return total

# Function calls
print("Sum of 1, 2, 3:", calculate_sum(1, 2, 3))
print("Sum of 5, 10, 15, 20:", calculate_sum(5, 10, 15, 20))
print("Sum of no numbers:", calculate_sum())
```

```python
#16
def find_max(*numbers):
    if len(numbers) == 0:
        return "No numbers provided"

    maximum = numbers[0]
    for num in numbers:
        if num > maximum:
            maximum = num
    return maximum

# Function calls
print("Maximum of 3, 7, 2:", find_max(3, 7, 2))
print("Maximum of 10, 25, 5, 15:", find_max(10, 25, 5, 15))
print("Maximum of single value:", find_max(42))
print("Maximum with no arguments:", find_max())
```

```python
#19
def show_values(*args, **kwargs):
    print("Positional arguments (*args):")
    for value in args:
        print(value)

    print("\nKeyword arguments (**kwargs):")
    for key, value in kwargs.items():
        print(f"{key}: {value}")

# Function call
show_values(10, 20, 30, name="Aarya", age=20, course="Python")
```

```python
#20
def average_sales(*sales):
    if len(sales) == 0:
        return "No sales data provided"

    total = 0
    for amount in sales:
        total += amount

    average = total / len(sales)
    return average

# Function calls
print("Average sales (3 branches):", average_sales(25000, 30000, 28000))
print("Average sales (5 branches):", average_sales(20000, 22000, 26000, 24000, 30000))
print("Average sales (no branches):", average_sales())
```

```python
#26
def expense_tracker():
    total_expense = 0   # Enclosing variable

    def add_expense(amount):
        nonlocal total_expense
        total_expense += amount
        print("Added expense:", amount)
        print("Total expenses so far:", total_expense)

    # Adding expenses
    add_expense(500)
    add_expense(1200)
    add_expense(300)
    add_expense(700)

# Function call
expense_tracker()
```

```python
#27
def concatenate_strings(*strings):
    result = ""
    for text in strings:
        result += text
    return result

# Function calls
print(concatenate_strings("Hello", " ", "World"))
print(concatenate_strings("Python", " ", "is", " ", "fun"))
print(concatenate_strings("A", "B", "C", "D"))
```

```python
#29
def order_details(*items, **prices):
    print("Order Items:")
    for item in items:
        print("-", item)

    print("\nItem Prices:")
    for item, price in prices.items():
        print(f"{item}: ₹{price}")

# Function call
order_details(
    "Laptop", "Mouse", "Keyboard",
    Laptop=55000,
    Mouse=500,
    Keyboard=1500
)
```

```python
#30
def salary_calculator(base_salary):
    total_salary = base_salary   # Enclosing variable

    def add_incentive(incentive):
        nonlocal total_salary
        total_salary += incentive
        print("Incentive added:", incentive)
        print("Total salary so far:", total_salary)

    # Adding incentives
    add_incentive(5000)
    add_incentive(3000)
    add_incentive(2000)

    print("Final Total Salary:", total_salary)

# Function call
salary_calculator(30000)
```
