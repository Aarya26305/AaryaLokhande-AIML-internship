```python
# 5. Remove stopwords from a given text using NLTK stopwords list
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

# Download required resources (run once)
nltk.download('punkt')
nltk.download('stopwords')

# Given text
text = "Python is easy to learn and it is widely used in data science"

# Tokenize text into words
words = word_tokenize(text)

# Load English stopwords
stop_words = set(stopwords.words('english'))

# Remove stopwords
filtered_words = [w for w in words if w.lower() not in stop_words]

# Print result
print(filtered_words)
```

```python
# 6. Perform stemming on a list of words using the PorterStemmer and print the output.
import nltk
from nltk.stem import PorterStemmer

# Initialize PorterStemmer
ps = PorterStemmer()

# List of words
words = ["running", "runs", "ran", "easily", "fairly"]

# Perform stemming
stemmed_words = [ps.stem(word) for word in words]

# Print output
print("Original words:", words)
print("Stemmed words:", stemmed_words)
```

```python
# 7. Perform lemmatization on the same list of words using WordNetLemmatizer and compare results with stemming.
import nltk
from nltk.stem import PorterStemmer, WordNetLemmatizer

# Download required resource (run once)
nltk.download('wordnet')

# Initialize stemmer and lemmatizer
ps = PorterStemmer()
lemmatizer = WordNetLemmatizer()

# Same list of words
words = ["running", "runs", "ran", "easily", "fairly"]

# Stemming
stemmed_words = [ps.stem(word) for word in words]

# Lemmatization
lemmatized_words = [lemmatizer.lemmatize(word) for word in words]

# Print comparison
print("Original words :", words)
print("Stemmed words  :", stemmed_words)
print("Lemmatized words:", lemmatized_words)
```

```python
# 8. Apply Part of Speech (POS) tagging to a given sentence and print each word with its tag.
import nltk
from nltk.tokenize import word_tokenize
from nltk import pos_tag

# Download required resources (run once)
nltk.download('punkt')
nltk.download('averaged_perceptron_tagger')

# Given sentence
sentence = "The quick brown fox jumps over the lazy dog"

# Tokenize sentence
words = word_tokenize(sentence)

# Apply POS tagging
pos_tags = pos_tag(words)

# Print each word with its POS tag
for word, tag in pos_tags:
    print(word, ":", tag)
```

```python
# 9. Perform frequency distribution on a paragraph and print the top 10 most frequent words.
import nltk
from nltk.tokenize import word_tokenize
from nltk.probability import FreqDist

# Download required resource (run once)
nltk.download('punkt')

# Given paragraph
paragraph = """
Natural Language Processing enables computers to understand human language.
It is widely used in applications like chatbots, search engines, and voice assistants.
"""

# Tokenize paragraph
words = word_tokenize(paragraph.lower())

# Remove punctuation and non-alphabetic words
words = [word for word in words if word.isalpha()]

# Create frequency distribution
freq_dist = FreqDist(words)

# Print top 10 most frequent words
print("Top 10 most frequent words:")
for word, freq in freq_dist.most_common(10):
    print(word, ":", freq)
```

```python
# 12. Create a function that accepts text input and returns a list of nouns and verbs separately using POS tagging.

import nltk
from nltk.tokenize import word_tokenize
from nltk import pos_tag

nltk.download('punkt')
nltk.download('averaged_perceptron_tagger')

def extract_nouns_verbs(text):
    tokens = word_tokenize(text)
    tagged_words = pos_tag(tokens)

    nouns = []
    verbs = []

    for word, tag in tagged_words:
        if tag.startswith('NN'):      # Nouns
            nouns.append(word)
        elif tag.startswith('VB'):    # Verbs
            verbs.append(word)

    return nouns, verbs


sentence = "The student writes code and learns Python quickly"
nouns, verbs = extract_nouns_verbs(sentence)

print("Nouns:", nouns)
print("Verbs:", verbs)
```

```python
# 13. Count the number of adjectives and adverbs present in a given text.
import nltk
from nltk.tokenize import word_tokenize
from nltk import pos_tag
nltk.download('punkt')
nltk.download('averaged_perceptron_tagger')

def count_adjectives_adverbs(text):
    tokens = word_tokenize(text)
    tagged_words = pos_tag(tokens)

    adj_count = 0
    adv_count = 0

    for word, tag in tagged_words:
        if tag.startswith('JJ'):   # Adjectives
            adj_count += 1
        elif tag.startswith('RB'): # Adverbs
            adv_count += 1

    return adj_count, adv_count

sentence = "The very smart student quickly solved the extremely difficult problem"
adjectives, adverbs = count_adjectives_adverbs(sentence)

print("Number of adjectives:", adjectives)
print("Number of adverbs:", adverbs)
```

```python
# 14. Tokenize a paragraph and find the average length of words using NLTK functions.

import nltk
from nltk.tokenize import word_tokenize, sent_tokenize
nltk.download('punkt')

def average_word_length(paragraph):
    words = word_tokenize(paragraph)
    
    words = [word for word in words if word.isalpha()]
    
    avg_length = sum(len(word) for word in words) / len(words)
    
    return avg_length

paragraph = "Natural Language Processing with Python is very interesting. It helps analyze text efficiently."
avg_len = average_word_length(paragraph)
print("Average word length:", round(avg_len, 2))
```

```python
# 15. Write a program to find all unique words from a text using tokenization and set().

import nltk
from nltk.tokenize import word_tokenize

# Download resources (run once)
nltk.download('punkt')

def unique_words(text):
    words = word_tokenize(text)
    
    words = [word.lower() for word in words if word.isalpha()]
    
    unique = set(words)
    
    return unique

text = "Python is easy to learn. Python is powerful and easy to use."
unique = unique_words(text)
print("Unique words:", unique)
print("Number of unique words:", len(unique))
```

```python
# 16. Compare and print the difference between stemming and lemmatization for a list of 10 verbs.

import nltk
from nltk.stem import PorterStemmer, WordNetLemmatizer

nltk.download('wordnet')
nltk.download('omw-1.4')

stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

verbs = ["running", "ate", "flying", "swimming", "written", 
         "studying", "drove", "played", "seeing", "went"]

stemmed_verbs = [stemmer.stem(verb) for verb in verbs]
lemmatized_verbs = [lemmatizer.lemmatize(verb, pos='v') for verb in verbs]

print("Original verbs:     ", verbs)
print("Stemmed verbs:      ", stemmed_verbs)
print("Lemmatized verbs:   ", lemmatized_verbs)
```
