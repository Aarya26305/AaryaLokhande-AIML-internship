```python
#5
num = int(input("Enter number: "))
rev = 0

while num > 0:
    rev = rev * 10 + num % 10
    num //= 10

print("Reversed number:", rev)
```

```python
#6
num = int(input("Enter number: "))
count = 0

while num != 0:
    count += 1
    num //= 10

print("Number of digits:", count)
```

```python
#7
num = int(input("Enter number: "))
fact = 1

while num > 0:
    fact *= num
    num -= 1

print("Factorial =", fact)
```

```python
#9
num = int(input("Enter number: "))
temp = num
rev = 0

while temp > 0:
    rev = rev * 10 + temp % 10
    temp //= 10

if num == rev:
    print("Palindrome")
else:
    print("Not Palindrome")
```
