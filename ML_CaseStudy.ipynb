```python
# Q2: Use info() and describe() to get a comprehensive overview of the dataset
# Identify data types, missing values, and statistical summary
# Print the column names using columns attribute and shape of the dataset


# Dataset overview
print("Dataset Info:")
df.info()

print("\nStatistical Summary:")
print(df.describe())

# Column names
print("\nColumn Names:")
print(df.columns)

# Shape of dataset
print("\nDataset Shape:")
print(df.shape)
```

```python
# Q3: Introduce 10% missing values randomly in Rating and Review columns
# Detect missing values and display the count of missing values per column
# Fill missing ratings with the median rating and missing reviews with "No review provided"

import numpy as np


df.loc[df.sample(frac=0.1).index, "Rating"] = np.nan
df.loc[df.sample(frac=0.1).index, "Review"] = np.nan


print("Missing values:")
print(df.isnull().sum())


df["Rating"] = df["Rating"].fillna(df["Rating"].median())
df["Review"] = df["Review"].fillna("No review provided")

# Check again
print("\nAfter filling missing values:")
print(df.isnull().sum())
```

```python
# Q4: Create a new column 'TotalAmount' by multiplying Quantity and Price
# Add another column 'DiscountedPrice' that applies 10% discount on orders above $100
# Use conditional selection to filter orders where TotalAmount> $150


df["TotalAmount"] = df["Quantity"] * df["Price"]
df["DiscountedPrice"] = df["TotalAmount"]
df.loc[df["TotalAmount"] > 10000, "DiscountedPrice"] = df["TotalAmount"] * 0.9
filtered_orders = df[df["TotalAmount"] > 150]
print("Filtered orders (TotalAmount > 150):")
filtered_orders
```

```python
# Q5: Select only the columns: Product, Category, TotalAmount, and Rating
# Filter and display all orders from the 'Electronics' category with Rating >= 4
# Select required columns

selected_columns = df[["Product", "Category", "TotalAmount", "Rating"]]


electronics_high_rating = selected_columns[
    (selected_columns["Category"] == "Electronics") &
    (selected_columns["Rating"] >= 4)
]


electronics_high_rating
```

```python
# Q6: Use boolean indexing to find all orders where:
# - Quantity > 2 AND Price < 50 OR Rating == 5
# Display the count of such orders



filtered_orders = df[
    ((df["Quantity"] > 2) & (df["Price"] < 50)) |
    (df["Rating"] == 5)
]


print("Number of matching orders:", filtered_orders.shape[0])
```

```python
# Q7: Sort the entire DataFrame by OrderDate (descending) and then by TotalAmount (descending)
# Create a new column 'SalesRank' that ranks products by TotalAmount in descending order
# Sort by OrderDate (descending) and TotalAmount (descending)
df_sorted = df.sort_values(
    by=["OrderDate", "TotalAmount"],
    ascending=[False, False]
)

# Create SalesRank column (higher TotalAmount = higher rank)
df_sorted["SalesRank"] = df_sorted["TotalAmount"].rank(
    method="dense",
    ascending=False
)

# Display result
print(df_sorted[["OrderID", "Product", "OrderDate", "TotalAmount", "SalesRank"]])

```

```python
# Q8: Find the top 5 best-selling products based on total quantity sold
# Display the product name and total quantity in descending order

product_sales = df.groupby("Product")["Quantity"].sum()
top_5_products = product_sales.sort_values(ascending=False).head(5)
print(top_5_products)
```

```python
# Q9: Group the data by 'Category' and calculate:
# - Total sales (sum of TotalAmount)
# - Average rating
# - Number of orders
# Display the results sorted by total sales

category_summary = (
    df.groupby("Category")
      .agg(
          TotalSales=("TotalAmount", "sum"),
          AverageRating=("Rating", "mean"),
          NumberOfOrders=("OrderID", "count")
      )
      .sort_values(by="TotalSales", ascending=False)
)
category_summary
```

```python
# Q10: Create a pivot table showing average Price for each Category and Rating combination
# Use groupby with multiple columns (Category, Product) and aggregate multiple functions

# Pivot table for average Price by Category and Rating
pivot_table = pd.pivot_table(
    df,
    values="Price",
    index="Category",
    columns="Rating",
    aggfunc="mean"
)

pivot_table
```

```python
# Group by Category and Product with multiple aggregations
grouped_data = (
    df.groupby(["Category", "Product"]).agg(
          TotalQuantity=("Quantity", "sum"),
          AvgPrice=("Price", "mean"),
          TotalSales=("TotalAmount", "sum")
      )
)

print(grouped_data)
```

```python
# Q16: Create a correlation matrix between Quantity, Price, Rating, and TotalAmount using NumPy
# Calculate the covariance matrix
# Perform linear algebra operations: determinant and inverse of correlation matrix

import numpy as np
data = df[["Quantity", "Price", "Rating", "TotalAmount"]].to_numpy()
correlation_matrix = np.corrcoef(data, rowvar=False)
print("Correlation Matrix:\n", correlation_matrix)
covariance_matrix = np.cov(data, rowvar=False)
print("\nCovariance Matrix:\n", covariance_matrix)
determinant = np.linalg.det(correlation_matrix)
print("\nDeterminant of Correlation Matrix:", determinant)
inverse_matrix = np.linalg.inv(correlation_matrix)
print("\nInverse of Correlation Matrix:\n", inverse_matrix)
```

```python
# Q18: Create a random array of 1000 sales values
# Use NumPy to flatten, reshape into (100, 10), and then transpose
# Compare memory usage between original, copy, and view


import numpy as np
sales = np.random.randint(100, 5000, 1000)

print("Original array shape:", sales.shape)
flat_array = sales.flatten()

reshaped_array = sales.reshape(100, 10)
transposed_array = reshaped_array.T

# Display shapes
print("Flattened shape:", flat_array.shape)
print("Reshaped shape:", reshaped_array.shape)
print("Transposed shape:", transposed_array.shape)
print("\nMemory Usage (in bytes):")
print("Original array:", sales.nbytes)
print("Flattened (copy):", flat_array.nbytes)
print("Reshaped (view):", reshaped_array.nbytes)
print("Transposed (view):", transposed_array.nbytes)
print("\nMemory Sharing:")
print("Reshaped shares memory:", reshaped_array.base is sales)
print("Transposed shares memory:", transposed_array.base is reshaped_array)
```

```python
# Q19: Create a line plot showing total sales trend over months
# Customize with: title, xlabel, ylabel, grid, legend
# Use different line styles, colors, and markers

import pandas as pd
import matplotlib.pyplot as plt
df["OrderDate"] = pd.to_datetime(df["OrderDate"])
df["Month"] = df["OrderDate"].dt.to_period("M")
monthly_sales = df.groupby("Month")["TotalAmount"].sum()
months = monthly_sales.index.astype(str)
sales = monthly_sales.values
plt.figure()
plt.plot(months, sales, linestyle='--', marker='o', color='blue', label='Total Sales')
plt.title("Monthly Total Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales Amount")
plt.grid(True)
plt.legend()


plt.show()
```
